```python
import pandas as pd
```

```python
player_awards = pd.read_csv("../cleaned_datasets/awards_players_cleaned.csv")
players = pd.read_csv("../cleaned_datasets/players_cleaned.csv")
coaches = pd.read_csv("../cleaned_datasets/coaches_cleaned.csv")
player_teams = pd.read_csv("../cleaned_datasets/players_teams_cleaned.csv")
postseason_matchups = pd.read_csv("../cleaned_datasets/series_post_cleaned.csv")
postseason_team_record = pd.read_csv("../cleaned_datasets/teams_post_cleaned.csv")
teams = pd.read_csv("../cleaned_datasets/teams_cleaned.csv")


```

```python
players.rename(columns={"bioID": "playerID"}, inplace=True)


df = pd.merge(player_teams, player_awards, on=["playerID", "year"], how="left")
df = pd.merge(df, teams, on=["year", "tmID"], how="left", suffixes=('', '_team'))
df = pd.merge(df, coaches, on=["year", "tmID"], how="left", suffixes=('_team', '_coach'))
df.rename(columns={'post_wins': 'post_wins_coach','post_losses': 'post_losses_coach'}, inplace=True)

df = pd.merge(df, players, on=["playerID"], how="left")
df = pd.merge(df, postseason_team_record, on=["year", "tmID"], how="left", suffixes=('', '_postseason'))
df.rename(columns={'W': 'post_season_team_W','L': 'post_season_team_L'}, inplace=True)

df.drop_duplicates(inplace=True)
df['playoff'] = df['playoff'].apply(lambda x: 1 if x == 'Y' else 0)


print(df.head())
print(df.shape)
columns = df.columns
print(columns)
df.to_csv('../datasetMerger/df_merge.csv', index=False)
```
