```python
import pandas as pd
```

```python
player_teams = pd.read_csv("basketballPlayoffs/players_teams.csv")
```

```python
# Remover as colunas especificadas: 'lgID', 'oRebounds', 'dRebounds', 'PostoRebounds', 'PostdRebounds'
columns_to_drop = ['lgID', 'oRebounds', 'dRebounds', 'PostoRebounds', 'PostdRebounds']
df_cleaned = player_teams.drop(columns=columns_to_drop)

# Mostrar as primeiras linhas do DataFrame atualizado para verificar
df_cleaned.head()
```

```python
df_cleaned['total_stats_regular'] = df_cleaned[['points', 'rebounds', 'assists', 'steals', 'blocks']].sum(axis=1)
df_cleaned['total_stats_post'] = df_cleaned[['PostPoints', 'PostRebounds', 'PostAssists', 'PostSteals', 'PostBlocks']].sum(axis=1)

columns_to_drop = ['points', 'rebounds', 'assists', 'steals', 'blocks', 'PostPoints', 'PostRebounds', 'PostAssists', 'PostSteals', 'PostBlocks']
df_cleaned = df_cleaned.drop(columns=columns_to_drop)
df_cleaned[['total_stats_regular', 'total_stats_post']].head()
```

```python
# Subtrair as colunas turnovers, PF, dq do total_stats_regular
df_cleaned['total_stats_regular'] = df_cleaned['total_stats_regular'] - df_cleaned[['turnovers', 'PF', 'dq']].sum(axis=1)

# Subtrair as colunas PostTurnovers, PostPF, PostDQ do total_stats_post
df_cleaned['total_stats_post'] = df_cleaned['total_stats_post'] - df_cleaned[['PostTurnovers', 'PostPF', 'PostDQ']].sum(axis=1)

columns_to_drop = ['turnovers', 'PF', 'dq', 'PostTurnovers', 'PostPF', 'PostDQ']
df_cleaned = df_cleaned.drop(columns=columns_to_drop)

# Mostrar as primeiras linhas do DataFrame atualizado com as colunas ajustadas
df_cleaned[['total_stats_regular', 'total_stats_post']].head()
```

```python
# Creating the calculated fields and removing specified columns.

# Calculating twoAttempted, twoMade, and twoFailed (2-point attempts, made, and failed)

df_cleaned['twoAttempted'] = df_cleaned['fgAttempted'] - df_cleaned['threeAttempted']
df_cleaned['twoMade'] = df_cleaned['fgMade'] - df_cleaned['threeMade']
df_cleaned['twoFailed'] = df_cleaned['twoAttempted'] - df_cleaned['twoMade']

# Calculating PosttwoAttempted, PosttwoMade, and PosttwoFailed
df_cleaned['PosttwoAttempted'] = df_cleaned['PostfgAttempted'] - df_cleaned['PostthreeAttempted']
df_cleaned['PosttwoMade'] = df_cleaned['PostfgMade'] - df_cleaned['PostthreeMade']
df_cleaned['PosttwoFailed'] = df_cleaned['PosttwoAttempted'] - df_cleaned['PosttwoMade']

# Calculating fgFailed, ftFailed, threeFailed, PostfgFailed, PostftFailed, PostthreeFailed
df_cleaned['fgFailed'] = df_cleaned['fgAttempted'] - df_cleaned['fgMade']
df_cleaned['ftFailed'] = df_cleaned['ftAttempted'] - df_cleaned['ftMade']
df_cleaned['threeFailed'] = df_cleaned['threeAttempted'] - df_cleaned['threeMade']

df_cleaned['PostfgFailed'] = df_cleaned['PostfgAttempted'] - df_cleaned['PostfgMade']
df_cleaned['PostftFailed'] = df_cleaned['PostftAttempted'] - df_cleaned['PostftMade']
df_cleaned['PostthreeFailed'] = df_cleaned['PostthreeAttempted'] - df_cleaned['PostthreeMade']

# Dropping the requested columns
columns_to_drop = ['fgAttempted', 'ftAttempted', 'threeAttempted', 'PostfgAttempted', 'PostftAttempted', 'PostthreeAttempted', 'PosttwoAttempted', 'twoAttempted']
df_cleaned = df_cleaned.drop(columns=columns_to_drop)

# Display the first few rows of the cleaned dataframe to verify
df_cleaned.head()
```

```python
# Dropping the requested columns
columns_to_drop = ['PostfgMade','PostftMade','PostthreeMade','fgMade','ftMade','threeMade','twoMade','twoFailed','PosttwoMade','PosttwoFailed','fgFailed','ftFailed','threeFailed','PostfgFailed','PostftFailed','PostthreeFailed']
df_cleaned = df_cleaned.drop(columns=columns_to_drop)
```

```python
# Exportar o dataframe atualizado para um arquivo CSV
df_cleaned.to_csv('palyers_score.csv', index=False)
```
