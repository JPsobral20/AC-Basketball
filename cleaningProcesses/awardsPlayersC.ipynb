# Dataset Cleaning for Player Awards

```python
import pandas as pd
player_awards = pd.read_csv("../basketballPlayoffs/awards_players.csv")
player_awards.head()
```

# Check For Null Values

```python
null = player_awards.isnull().values.any()
nan = player_awards.isna().values.any()

print(null)
print(nan)

# No Missing Values detected so no need of cleaning
```

# Dropping Unique Value Columns

```python
for name in player_awards:
    for col in player_awards.columns:
        if len(player_awards[col].unique()) == 1:
            player_awards.drop(col, inplace=True, axis=1)
    print(f"Processed {name}: Remaining columns: {player_awards.columns}")
```

# Dealing with Redundancy

```python

row_count = player_awards.shape[0]
player_awards.drop_duplicates(inplace=True)
afterDups_count = player_awards.shape[0]
rows_deleted = row_count - afterDups_count
print(f"Number of rows deleted: {rows_deleted}")
```

# Dealing With Outliers

```python
award_list = player_awards["award"].unique().tolist()
print(award_list)

## Kim Perrot Sportsmanship and Kim Perrot Sportmanship Award can all be converted into a single one (trophy renamed for year 10)

player_awards['award'] = player_awards['award'].replace('Kim Perrot Sportsmanship', 'Kim Perrot Sportsmanship Award')
print(player_awards['award'].unique())



```

# Create Clean CSV

```python
player_awards.to_csv('../cleaned_datasets/awards_players_cleaned.csv', index=False)
```
